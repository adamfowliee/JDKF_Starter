```python
import pandas as pd
```

```python
df = pd.read_excel("./datasets/world_gdps.xlsx")
```

```python
cols = df.iloc[1]
gdp_df = df[2:]
gdp_df.columns = cols

no_data_df =  gdp_df[gdp_df['GDP, current prices (Billions of U.S. dollars)'] == 'no data']
gdp_df = gdp_df[gdp_df['GDP, current prices (Billions of U.S. dollars)'] != 'no data']

gdp_df.head()
```

```python
gdp_df.sort_values("GDP, current prices (Billions of U.S. dollars)", ascending=False).to_excel("./datasets/sorted_gdp.xlsx")
```
